## 1. La legge dei grandi numeri

La legge dei grandi numeri è un teorema fondamentale della teoria della probabilità che indica che se ripetiamo molte volte (tendendo all'infinito) lo stesso esperimento, la frequenza di un certo evento tende ad essere costante.

Vale a dire, **la legge dei grandi numeri indica che se lo stesso test viene eseguito ripetutamente (ad esempio, lancio di una moneta), la frequenza con cui si ripeterà un determinato evento (testa/croce) si avvicinerà a una costante. Questa a sua volta sarà la probabilità che questo evento si verifichi.**

Fonte: https://it.economy-pedia.com/11038376-law-of-the-big-numbers

- Scrivere una funzione che simula il lancio di una moneta per N volte e restituisce la frequenza % di volte che è uscita testa.
- Calcolare la frequenza % con la grandezze del campione (numero di lanci) che varia 100 volte da 10 a 20000. Ottengo quindi 100 valori.
- Fare il plot del Numero medio di teste (y) rispetto alla grandezza del campione (x)







```python
def simula_lanci(n):
    lanci = np.random.randint(0, 2, size=n)  
    return lanci.mean() * 100

dimensioni = np.linspace(10,2000,100,dtype=int)
frequenze = [round(simula_lanci(n),2) for n in dimensioni ]


plt.plot(dimensioni,frequenze)
plt.axhline(50, color="red", linestyle="--")



```

## 2. Analisi del Dataset Titanic
- Quante righe e colonne ha il dataset?
- Controlla quanti valori mancanti ci sono per colonna
- Riempi i valori mancanti nella colonna 'Embarked' con il valore più frequente
- Controlla se ci sono righe duplicate
- Calcola l'età media dei passeggeri per ogni classe (`Pclass`), se ci sono valori mancanti di età nella colonna 'Age' riempili con il valore medio
- Visualizza la distribuzione dell'età per classe
- Visualizza la distribuzione dell'età per classe dividendo e mostrando insieme i dati di uomini e donne

```python
# 📦 Caricamento del dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()

#righe e colonne 
df.shape

#valori mancanti 
df.isnull().sum()

#riempre valori mancanti
più_frequente = df["Embarked"].mode()
df["Embarked"] = df["Embarked"].fillna(più_frequente)

#righe duplicate
df.duplicated().sum()

#media per classe
valore_medio = df['Age'].mean()
df['Age'] = df['Age'].fillna(valore_medio)
df.groupby('Pclass')['Age'].mean()

#distribuzione per classe
sns.boxplot(data=df, x="Pclass",y="Age")

#distribuzione dividendo per uomini e donne
sns.boxplot(data=df, x="Pclass",y="Age", hue='Sex')
```

## 3. Analisi del Dataset Iris

- Esplora la distribuzione delle specie (conta quanti campioni ci sono per specie)
- Calcola la lunghezza e la larghezza media dei petali per specie
- Visualizza le dimensioni dei petali per specie (scatterplot)
- Crea una nuova colonna per l'area del petalo e analizzala
- Grafico della distribuzione dell'area del petalo per specie (boxplot)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)


conteggio_specie = df['species'].value_counts()
print(f"distribuzione delle specie nel dataset:{conteggio_specie}")

medie_petali = df.groupby('species')[['petal_length', 'petal_width']].mean()
print(f"medie delle dimensioni dei petali per ogni specie:{medie_petali}")

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species')
plt.title('dimensioni dei Petali per Specie di Iris')
plt.xlabel('lunghezza Petalo (cm)')
plt.ylabel('larghezza Petalo (cm)')
plt.show()

df['petal_area'] = df['petal_length'] * df['petal_width']
analisi_area = df.groupby('species')['petal_area'].describe()
print(f"analisi statistica dell'area del petalo per specie: {analisi_area[['count', 'mean', 'min', 'max']]}")

plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='species', y='petal_area')
plt.title('distribuzione dell\'Area del Petalo per Specie')
plt.xlabel('specie')
plt.ylabel('area del Petalo ($cm^2$)')
plt.show()

```
